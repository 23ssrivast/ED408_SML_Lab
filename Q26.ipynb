```python
import numpy as np

train = np.loadtxt("Datasets/pendigits.tra", delimiter=",")
test = np.loadtxt("Datasets/pendigits.tes", delimiter=",")
Xtr = train[:, :16] / 100
Xte = test[:, :16] / 100
ttr = (train[:, 16].astype(int) % 2).reshape(-1, 1)
tte = test[:, 16].astype(int) % 2
n = len(Xtr)

print("Training samples:", n, " Testing samples:", len(Xte))

np.random.seed(1)
W1 = np.random.randn(16, 24) * 0.3
b1 = np.zeros(24)
W2 = np.random.randn(24, 1) * 0.2
b2 = np.zeros(1)
lr = 0.3
epochs = 1000


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def forward(X):
    h = np.tanh(np.dot(X, W1) + b1)
    p = sigmoid(np.dot(h, W2) + b2)
    return h, p


def cost(p, t):
    return -np.mean(t * np.log(p) + (1 - t) * np.log(1 - p))


for i in range(epochs):
    h, p = forward(Xtr)
    if i % 100 == 0:
        print("Epoch", i, " Loss =", round(cost(p, ttr), 4))
    d2 = (p - ttr) / n
    d1 = np.dot(d2, W2.T) * (1 - h ** 2)
    W2 = W2 - lr * np.dot(h.T, d2)
    b2 = b2 - lr * d2.sum(axis=0)
    W1 = W1 - lr * np.dot(Xtr.T, d1)
    b1 = b1 - lr * d1.sum(axis=0)

h, p = forward(Xtr)
print("Final Loss:", round(cost(p, ttr), 4))

h, p = forward(Xte)
pred = (p.reshape(-1) >= 0.5).astype(int)
print("Test Accuracy (odd vs even digit):", round((pred == tte).mean(), 4))
```
