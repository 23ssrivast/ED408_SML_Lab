```python
import gzip
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models


def load_images(name):
    f = gzip.open("Datasets/" + name, "rb")
    data = np.frombuffer(f.read(), dtype=np.uint8, offset=16)
    f.close()
    return data.reshape(-1, 28, 28, 1)


def load_labels(name):
    f = gzip.open("Datasets/" + name, "rb")
    data = np.frombuffer(f.read(), dtype=np.uint8, offset=8)
    f.close()
    return data


Xtr = load_images("train-images-idx3-ubyte.gz") / 255.0
ytr = load_labels("train-labels-idx1-ubyte.gz")
Xte = load_images("t10k-images-idx3-ubyte.gz") / 255.0
yte = load_labels("t10k-labels-idx1-ubyte.gz")

print("Training images:", Xtr.shape, " Testing images:", Xte.shape)

tf.keras.utils.set_random_seed(42)

model = models.Sequential()
model.add(layers.Input((28, 28, 1)))
model.add(layers.Conv2D(16, 3, activation="relu"))
model.add(layers.MaxPooling2D())
model.add(layers.Conv2D(32, 3, activation="relu"))
model.add(layers.MaxPooling2D())
model.add(layers.Flatten())
model.add(layers.Dense(10, activation="softmax"))

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.fit(Xtr, ytr, epochs=3, batch_size=128, verbose=2)

loss, acc = model.evaluate(Xte, yte, verbose=0)
print("Test Loss:", round(loss, 4))
print("Test Accuracy:", round(acc, 4))
```
