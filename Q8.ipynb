```python
import numpy as np


def f(x):
    return x ** 2 + 10 * np.sin(x)


np.random.seed(1)
x = 8.0
best = x
T = 5.0
cooling = 0.998

print("Starting point x =", x, " f(x) =", round(f(x), 5))

for i in range(4000):
    new = x + np.random.normal(0, 0.8)
    new = max(-10, min(10, new))
    diff = f(new) - f(x)
    if diff < 0 or np.random.rand() < np.exp(-diff / T):
        x = new
    if f(x) < f(best):
        best = x
    T = T * cooling
    if i % 500 == 0:
        print("Iteration", i, " Temperature =", round(T, 4), " Best f(x) =", round(f(best), 5))

print("Final temperature:", round(T, 5))
print("Best x =", round(best, 5))
print("Minimum f(x) =", round(f(best), 5))
```
