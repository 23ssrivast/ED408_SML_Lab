```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

df = pd.read_csv("Datasets/bank-full.csv", sep=";")
X = df[["age", "balance", "campaign", "pdays", "previous"]].values
y = (df["y"] == "yes").astype(int).values

k = 5
np.random.seed(1)
idx = np.random.permutation(len(X))
folds = np.array_split(idx, k)

scores = []
for i in range(k):
    test_idx = folds[i]
    train_idx = []
    for j in range(k):
        if j != i:
            train_idx.extend(folds[j])
    train_idx = np.array(train_idx)
    model = DecisionTreeClassifier(max_depth=6, random_state=42)
    model.fit(X[train_idx], y[train_idx])
    pred = model.predict(X[test_idx])
    acc = (pred == y[test_idx]).mean()
    scores.append(acc)
    print("Fold", i + 1, " Train size =", len(train_idx), " Test size =", len(test_idx), " Accuracy =", round(acc, 4))

print("Mean Accuracy:", round(np.mean(scores), 4))
print("Standard Deviation:", round(np.std(scores), 4))
```
