```python
import numpy as np

f = open("Datasets/pr1002.tsp")
text = f.read()
f.close()

lines = text.split("NODE_COORD_SECTION")[1].split("EOF")[0].strip().split("\n")
cities = []
for line in lines:
    parts = line.split()
    cities.append([float(parts[1]), float(parts[2])])
cities = np.array(cities)
n = len(cities)
print("Number of cities:", n)

dist = np.zeros((n, n))
for i in range(n):
    dist[i] = np.round(np.sqrt(((cities - cities[i]) ** 2).sum(axis=1)))

np.random.seed(1)
pop_size = 100
generations = 5000
pc = 0.9
pm = 0.3


def tour_length(tour):
    return int(dist[tour[:-1], tour[1:]].sum() + dist[tour[-1], tour[0]])


def select(pop, lengths):
    idx = np.random.randint(0, pop_size, 5)
    return pop[idx[np.argmin(lengths[idx])]]


def crossover(p1, p2):
    a, b = sorted(np.random.randint(0, n, 2))
    middle = p1[a:b]
    rest = p2[~np.isin(p2, middle)]
    return np.concatenate([rest[:a], middle, rest[a:]])


def mutate(tour):
    if np.random.rand() < pm:
        a, b = sorted(np.random.randint(0, n, 2))
        tour[a:b] = tour[a:b][::-1]
    return tour


pop = []
for i in range(pop_size):
    pop.append(np.random.permutation(n))

for g in range(generations):
    lengths = np.array([tour_length(t) for t in pop])
    if g % 500 == 0:
        print("Generation", g, " Best length =", lengths.min())
    new_pop = [pop[np.argmin(lengths)]]
    while len(new_pop) < pop_size:
        p1 = select(pop, lengths)
        p2 = select(pop, lengths)
        if np.random.rand() < pc:
            child = crossover(p1, p2)
        else:
            child = p1.copy()
        new_pop.append(mutate(child))
    pop = new_pop

lengths = np.array([tour_length(t) for t in pop])
best = pop[np.argmin(lengths)]
print("Best tour length after", generations, "generations:", lengths.min())
print("Cities visited:", len(set(best.tolist())))
print("First 12 cities of the tour:", (best[:12] + 1).tolist())
```
