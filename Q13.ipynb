```python
import numpy as np

np.random.seed(1)
n = 100
cost = np.random.randint(1, 21, (n, n, 2))
pher = np.ones((n, n, 2))

ants = 64
iterations = 70
alpha = 1
beta = 2
rho = 0.15
Q = 500

best_cost = np.inf
best_path = []

for it in range(iterations):
    paths = []
    totals = []
    for a in range(ants):
        r = 0
        c = 0
        path = []
        total = 0
        while r < n - 1 or c < n - 1:
            if c == n - 1:
                move = 1
            elif r == n - 1:
                move = 0
            else:
                p = pher[r, c] ** alpha / cost[r, c] ** beta
                if np.random.rand() < p[0] / p.sum():
                    move = 0
                else:
                    move = 1
            path.append((r, c, move))
            total = total + cost[r, c, move]
            if move == 0:
                c = c + 1
            else:
                r = r + 1
        paths.append(path)
        totals.append(total)

    k = int(np.argmin(totals))
    if totals[k] < best_cost:
        best_cost = totals[k]
        best_path = paths[k]

    pher = np.maximum(pher * (1 - rho), 0.05)
    for r, c, move in paths[k]:
        pher[r, c, move] = pher[r, c, move] + Q / totals[k]

    if it % 10 == 0:
        print("Iteration", it, " Best cost so far =", best_cost)

route = ""
for r, c, move in best_path:
    if move == 0:
        route = route + "E"
    else:
        route = route + "S"

print("Shortest path cost found:", best_cost)
print("Number of steps:", len(best_path))
print("Path from (0, 0) to (99, 99):", route)
```
