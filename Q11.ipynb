```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Lasso, Ridge, ElasticNet
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("Datasets/hour.csv")

X = df.drop(columns=["instant", "dteday", "casual", "registered", "cnt"])
y = df["cnt"]
names = list(X.columns)

s = int(0.8 * len(df))
Xtr, Xte = X[:s], X[s:]
ytr, yte = y[:s], y[s:]

scaler = StandardScaler()
Xtr = scaler.fit_transform(Xtr)
Xte = scaler.transform(Xte)

print("Training samples:", len(Xtr), " Testing samples:", len(Xte))

models = [("L1 (Lasso)", Lasso(alpha=1)), ("L2 (Ridge)", Ridge(alpha=1)), ("Elastic Net", ElasticNet(alpha=1, l1_ratio=0.5))]
coefs = {}

for name, m in models:
    m.fit(Xtr, ytr)
    pred = m.predict(Xte)
    rmse = np.sqrt(np.mean((yte - pred) ** 2))
    coefs[name] = m.coef_
    print(name, " RMSE =", round(rmse, 2), " Zero coefficients =", int((m.coef_ == 0).sum()))

print("Coefficients:")
print(pd.DataFrame(coefs, index=names).round(3))
```
