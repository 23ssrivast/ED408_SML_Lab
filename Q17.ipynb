```python
import numpy as np
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

train = np.loadtxt("Datasets/pendigits.tra", delimiter=",")
test = np.loadtxt("Datasets/pendigits.tes", delimiter=",")
Xtr, ytr = train[:, :16] / 100, train[:, 16].astype(int)
Xte, yte = test[:, :16] / 100, test[:, 16].astype(int)

print("Training samples:", len(Xtr), " Testing samples:", len(Xte))

bag = BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=80, random_state=42)
bag.fit(Xtr, ytr)
acc_bag = accuracy_score(yte, bag.predict(Xte))

rf = RandomForestClassifier(n_estimators=80, max_features="sqrt", random_state=42)
rf.fit(Xtr, ytr)
acc_rf = accuracy_score(yte, rf.predict(Xte))

print("Bagging Test Accuracy:", round(acc_bag, 4))
print("Random Forest Test Accuracy:", round(acc_rf, 4))

if acc_rf > acc_bag:
    print("Random Forest performed better by", round(acc_rf - acc_bag, 4))
else:
    print("Bagging performed better by", round(acc_bag - acc_rf, 4))
```
