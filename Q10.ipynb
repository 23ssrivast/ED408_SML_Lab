```python
import numpy as np


def rosenbrock(p):
    x = p[:, 0]
    y = p[:, 1]
    return (1 - x) ** 2 + 100 * (y - x ** 2) ** 2


np.random.seed(1)
n = 40
w = 0.7
c1 = 1.5
c2 = 1.5

pos = np.random.uniform(-3, 3, (n, 2))
vel = np.zeros((n, 2))
pbest = pos.copy()

for i in range(300):
    gbest = pbest[np.argmin(rosenbrock(pbest))].copy()
    r1 = np.random.rand(n, 2)
    r2 = np.random.rand(n, 2)
    vel = w * vel + c1 * r1 * (pbest - pos) + c2 * r2 * (gbest - pos)
    pos = np.clip(pos + vel, -3, 3)
    better = rosenbrock(pos) < rosenbrock(pbest)
    pbest[better] = pos[better]
    if i % 50 == 0:
        print("Iteration", i, " Best value =", rosenbrock(pbest).min())

gbest = pbest[np.argmin(rosenbrock(pbest))]
print("Best position:", np.round(gbest, 6))
print("Rosenbrock value:", rosenbrock(gbest.reshape(1, 2))[0])
```
