```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

df = pd.read_csv("Datasets/bank-full.csv", sep=";")
X = df[["age", "balance", "campaign", "pdays", "previous"]].values
y = (df["y"] == "yes").astype(int).values
n = len(X)

np.random.seed(1)
B = 40
scores = []

for b in range(B):
    idx = np.random.randint(0, n, n)
    oob = np.setdiff1d(np.arange(n), idx)
    model = DecisionTreeClassifier(max_depth=6, random_state=42)
    model.fit(X[idx], y[idx])
    acc = (model.predict(X[oob]) == y[oob]).mean()
    scores.append(acc)
    if b % 10 == 0:
        print("Bootstrap sample", b + 1, " Out-of-bag size =", len(oob), " Accuracy =", round(acc, 4))

scores = np.array(scores)
print("Number of bootstrap samples:", B)
print("Bootstrap Accuracy Estimate:", round(scores.mean(), 4))
print("Standard Deviation:", round(scores.std(), 4))
```
